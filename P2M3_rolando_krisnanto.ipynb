# Exploratory Data Analysis

Melakukan pengecekan, cleaning, dan perbaikan data sederhana sebelum masuk ke tahap Cek Validasi menggunakan Great Expectation

```python
# Read CSV File dan menampilkan 5 data teratas
df = pd.read_csv('data/P2M3_rolando_krisnanto_data_raw.csv')

df.head(5)
```

```python
# Cek Sturuktur Data
df.info()
```

```python
# Cek Data Duplikat
cek_dupl = df.duplicated().sum()

print ("Jumlah Data Duplikat : ", cek_dupl)
```

```python
# Ubah Tipe Data Kolom 'Date' menjadi Datetime
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')

df.info()

df.head(5)
```

```python
# Hapus row yang duplikat
df = df.drop_duplicates()

# Reset index agar urutan index mulai dari 0 lagi
df = df.reset_index(drop=True)

# Cek jumlah baris setelah drop data duplikat & cek data duplikat yang tersisa
print("Jumlah baris setelah hapus data duplikat:", len(df))
print("Jumlah data duplikat:", df.duplicated().sum())

df.head(5)
```

```python
# Buat Kolom Identifier untuk setiap baris yang diwakili oleh Transaction ID
df['Transaction_ID'] = range(1, len(df) + 1)

# Pindahkan Kolom Transaction_ID ke posisi paling kiri
## Ambil kolomnya dulu dan hapus dari dataframe
transaction_id = df.pop('Transaction_ID')

## masukkan lagi ke dataframe di posisi index ke-0
df.insert(0, 'Transaction_ID', transaction_id)

df.head(5)
```

# Great Expectation

Validasi data dengan library Great Expectation

```python
# Buat Data Context untuk Great Expectation
context = gx.get_context()
```

```python
# Membuat Data Source untuk pandas
data_source = context.data_sources.add_pandas("europe-bike-sales-source")

# Membuat Data Asset dari dataframe pandas
data_asset = data_source.add_dataframe_asset(name="europe-bike-sales-asset")

# Membuat Batch Definition (seluruh dataframe) untuk data asset
batch_definition = data_asset.add_batch_definition_whole_dataframe("bike-sales-batch")

# Ambil Batch, masukkan data frame yang sudah melalui EDA dan Cleaning
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})
```

```python
# Expectation 1 - To Be Unique
## untuk memvalidasi kolom Transaction_ID agar tidak ada duplikat
expectation = gx.expectations.ExpectColumnValuesToBeUnique(
    column="Transaction_ID"
)
result = batch.validate(expectation)
result
```

```python
# Expectation 2 - To Be Between (Min Value and Max Value)
## untuk memvalidasi kolom Customer_Age agar nilainya harus di antara 0 dan 90
expectation = gx.expectations.ExpectColumnValuesToBeBetween(
    column="Customer_Age", 
    min_value=0, 
    max_value=90
)
result = batch.validate(expectation)
result
```

```python
# Expectation 3 - To Be In Set
## untuk memvalidasi kolom Customer_Gender agar nilainya harus "M" atau "F"
expectation = gx.expectations.ExpectColumnValuesToBeInSet(
    column="Customer_Gender", 
    value_set=["M", "F"]
)
result = batch.validate(expectation)
result
```

```python
# Expectation 4 - To Be In Type List
## untuk memvalidasi kolom Date tipe datanya harus sesuai yang ditentukan, yaitu datetime64[ns]
expectation = gx.expectations.ExpectColumnValuesToBeInTypeList(
    column="Date", 
    type_list=["datetime64[ns]"]
)
result = batch.validate(expectation)
result
```

```python
# Expectation 5 - Column Pair Values A to Be Greater Than B 
## untuk memvalidasi kolom Unit_Price datanya harus lebih besar dari kolom Unit_Cost
expectation = gx.expectations.ExpectColumnPairValuesAToBeGreaterThanB(
    column_A="Unit_Price", 
    column_B="Unit_Cost",
    or_equal=True
)
result = batch.validate(expectation)
result
```

```python
# Expectation 6 - Value Lengths To Be Between
## untuk memvalidasi panjang nilai di kolom Product harus di antara 5 dan 75 karakte
expectation = gx.expectations.ExpectColumnValueLengthsToBeBetween(
    column="Product",
    min_value=5,
    max_value=75
)
result = batch.validate(expectation)
result
```

```python
# Expectation 7 - Table Row Count To Be Between
## untuk memvalidasi jumlah row di tabel harus diantara 100000 dan 120000
expectation = gx.expectations.ExpectTableRowCountToBeBetween(
    min_value=100000,
    max_value=120000
)
result = batch.validate(expectation)
result
```

Kumpulkan semua Expectations ke Expectation Suite, agar nanti bisa dialankan dalam 1 Checkpoint


```python
suite = context.suites.add(
    gx.ExpectationSuite(name="europe-bike-sales-suite")
)

suite.add_expectation(gx.expectations.ExpectColumnValuesToBeUnique(column="Transaction_ID"))
suite.add_expectation(gx.expectations.ExpectColumnValuesToBeBetween(column="Customer_Age", min_value=0, max_value=100))
suite.add_expectation(gx.expectations.ExpectColumnValuesToBeInSet(column="Customer_Gender", value_set=["M", "F"]))
suite.add_expectation(gx.expectations.ExpectColumnValuesToBeInTypeList(column="Date", type_list=["datetime64[ns]"]))
suite.add_expectation(gx.expectations.ExpectColumnPairValuesAToBeGreaterThanB(column_A="Unit_Price", column_B="Unit_Cost", or_equal=True))
suite.add_expectation(gx.expectations.ExpectColumnValueLengthsToBeBetween(column="Product", min_value=5, max_value=75))
suite.add_expectation(gx.expectations.ExpectTableRowCountToBeBetween(min_value=100000, max_value=120000))
```

Hubungkan "batch_definition" dengan "suite", lalu gabung keduanya sebelum dijalankan lewat Checkpoint

```python
validation_definition = context.validation_definitions.add(
    gx.ValidationDefinition(
        name="europe-bike-sales-validation",
        data=batch_definition,
        suite=suite,
    )
)
```

Bungkus Validation Definition diatas jadi satu kesatuan yaitu "Checkpoint", dan Jalankan Checkpointnya

```python
checkpoint = context.checkpoints.add(
    gx.Checkpoint(
        name="bike-sales-checkpoint",
        validation_definitions=[validation_definition],
    )
)
```

```python
checkpoint_result = checkpoint.run(batch_parameters={"dataframe": df})
checkpoint_result
```
